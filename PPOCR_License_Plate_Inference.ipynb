## 🛠️ STEP 6: Define Helper Functions for Visualization

```python
def extract_text_and_confidence(ocr_result: List) -> Tuple[str, float, int]:
    """
    Extract recognized text and average confidence from OCR result

    Args:
        ocr_result: OCR results from PaddleOCR

    Returns:
        Tuple of (text, average_confidence, num_detections)
    """
    # Check if result is empty or None
    if not ocr_result:
        return "No text detected", 0.0, 0

    # NEW FORMAT: Result is a list of dicts
    if isinstance(ocr_result, list) and len(ocr_result) > 0:
        if isinstance(ocr_result[0], dict):
            # New format: dict with 'rec_texts' and 'rec_scores'
            result_dict = ocr_result[0]
            texts = result_dict.get('rec_texts', [])
            confidences = result_dict.get('rec_scores', [])

            if not texts or len(texts) == 0:
                return "No text detected", 0.0, 0

            full_text = " ".join(texts)
            avg_confidence = np.mean(confidences) if confidences else 0.0
            num_detections = len(texts)

            return full_text, avg_confidence, num_detections

        # OLD FORMAT: List of lists [[box, (text, conf)], ...]
        elif len(ocr_result[0]) > 0 and isinstance(ocr_result[0][0], list):
            texts = []
            confidences = []

            for line in ocr_result[0]:
                text = line[1][0]
                confidence = line[1][1]
                texts.append(text)
                confidences.append(confidence)

            full_text = " ".join(texts)
            avg_confidence = np.mean(confidences) if confidences else 0.0
            num_detections = len(texts)

            return full_text, avg_confidence, num_detections

    return "No text detected", 0.0, 0


def draw_ocr_results(image_path: str, ocr_result: List, title: str = "", color: str = 'red') -> np.ndarray:
    """
    Draw OCR results on image with bounding boxes and text

    Args:
        image_path: Path to the image
        ocr_result: OCR results from PaddleOCR
        title: Title for the visualization
        color: Color for bounding boxes and text

    Returns:
        Annotated image as numpy array
    """
    # Read image
    img = cv2.imread(str(image_path))
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img_pil = Image.fromarray(img)
    draw = ImageDraw.Draw(img_pil)

    # Try to use a nice font
    try:
        font = ImageFont.truetype("DejaVuSans.ttf", 18)
        font_small = ImageFont.truetype("DejaVuSans.ttf", 14)
    except:
        try:
            font = ImageFont.truetype("arial.ttf", 18)
            font_small = ImageFont.truetype("arial.ttf", 14)
        except:
            font = ImageFont.load_default()
            font_small = ImageFont.load_default()

    # Check if result exists and is not empty
    if not ocr_result or len(ocr_result) == 0:
        return np.array(img_pil)

    # NEW FORMAT: Dict with rec_texts, rec_scores, rec_polys
    if isinstance(ocr_result[0], dict):
        result_dict = ocr_result[0]
        texts = result_dict.get('rec_texts', [])
        scores = result_dict.get('rec_scores', [])
        polys = result_dict.get('rec_polys', [])

        for idx, (text, confidence, poly) in enumerate(zip(texts, scores, polys)):
            if len(poly) == 0:
                continue

            # Draw bounding box
            box_points = [(int(point[0]), int(point[1])) for point in poly]
            draw.polygon(box_points, outline=color, width=3)

            # Draw text and confidence
            text_position = (int(poly[0][0]), int(poly[0][1]) - 30)
            label = f"#{idx+1}: {text}"
            conf_label = f"Conf: {confidence:.3f}"

            # Draw background rectangle for text
            try:
                bbox = draw.textbbox(text_position, label, font=font)
                draw.rectangle(bbox, fill='black')
            except:
                pass

            draw.text(text_position, label, fill=color, font=font)

            # Draw confidence below
            conf_position = (int(poly[0][0]), int(poly[0][1]) - 12)
            draw.text(conf_position, conf_label, fill='yellow', font=font_small)

    # OLD FORMAT: List of [box, (text, conf)]
    elif len(ocr_result[0]) > 0 and isinstance(ocr_result[0][0], list):
        for idx, line in enumerate(ocr_result[0]):
            box = line[0]
            text = line[1][0]
            confidence = line[1][1]

            # Draw bounding box
            box_points = [(int(point[0]), int(point[1])) for point in box]
            draw.polygon(box_points, outline=color, width=3)

            # Draw text and confidence
            text_position = (int(box[0][0]), int(box[0][1]) - 30)
            label = f"#{idx+1}: {text}"
            conf_label = f"Conf: {confidence:.3f}"

            # Draw background rectangle for text
            try:
                bbox = draw.textbbox(text_position, label, font=font)
                draw.rectangle(bbox, fill='black')
            except:
                pass

            draw.text(text_position, label, fill=color, font=font)

            # Draw confidence below
            conf_position = (int(box[0][0]), int(box[0][1]) - 12)
            draw.text(conf_position, conf_label, fill='yellow', font=font_small)

    return np.array(img_pil)
```

```python
# Process all test images
print("="*60)
print("📦 BATCH PROCESSING ALL TEST IMAGES")
print("="*60)

results_summary = []

for idx, img_path in enumerate(test_images, 1):
    print(f"\n🖼️ Processing {idx}/{len(test_images)}: {img_path.name}")

    # PPOCR v4
    start = time.time()
    result_v4 = ocr_v4.ocr(str(img_path))
    time_v4 = time.time() - start
    text_v4, conf_v4, num_v4 = extract_text_and_confidence(result_v4)

    # PPOCR v5
    start = time.time()
    result_v5 = ocr_v5.ocr(str(img_path))
    time_v5 = time.time() - start
    text_v5, conf_v5, num_v5 = extract_text_and_confidence(result_v5)

    # Store results
    results_summary.append({
        'Image': img_path.name,
        'v4_Text': text_v4,
        'v4_Confidence': conf_v4,
        'v4_Detections': num_v4,
        'v4_Time': time_v4,
        'v5_Text': text_v5,
        'v5_Confidence': conf_v5,
        'v5_Detections': num_v5,
        'v5_Time': time_v5
    })

    print(f"  ✅ v4: {text_v4[:50]}... | Conf: {conf_v4:.3f} | Time: {time_v4:.3f}s")
    print(f"  ✅ v5: {text_v5[:50]}... | Conf: {conf_v5:.3f} | Time: {time_v5:.3f}s")

print(f"\n✅ Batch processing completed! Processed {len(test_images)} images")
```

```python
# Performance comparison insights
print(f"\n{'='*60}")
print("💡 KEY INSIGHTS")
print(f"{'='*60}")

avg_conf_v4 = df_results['v4_Confidence'].mean()
avg_conf_v5 = df_results['v5_Confidence'].mean()
avg_time_v4 = df_results['v4_Time'].mean()
avg_time_v5 = df_results['v5_Time'].mean()

print(f"\n🎯 Confidence:")
if avg_conf_v5 > avg_conf_v4:
    improvement = ((avg_conf_v5 - avg_conf_v4) / avg_conf_v4) * 100
    print(f"  ✅ PPOCR v5 is MORE confident by {improvement:.2f}%")
else:
    decline = ((avg_conf_v4 - avg_conf_v5) / avg_conf_v4) * 100
    print(f"  ⚠️ PPOCR v4 is MORE confident by {decline:.2f}%")

print(f"\n⚡ Speed:")
if avg_time_v5 < avg_time_v4:
    speedup = ((avg_time_v4 - avg_time_v5) / avg_time_v4) * 100
    print(f"  ✅ PPOCR v5 is FASTER by {speedup:.2f}%")
else:
    slowdown = ((avg_time_v5 - avg_time_v4) / avg_time_v4) * 100
    print(f"  ⚠️ PPOCR v5 is SLOWER by {slowdown:.2f}%")

print(f"\n🎯 Recommendation:")
if avg_conf_v5 > avg_conf_v4 and avg_time_v5 < avg_time_v4:
    print(f"  ✅ Use PPOCR v5 - Better accuracy AND faster!")
elif avg_conf_v5 > avg_conf_v4:
    print(f"  ⚖️ Use PPOCR v5 - Better accuracy (slightly slower)")
elif avg_time_v5 < avg_time_v4:
    print(f"  ⚖️ Use PPOCR v5 - Faster (slightly lower accuracy)")
else:
    print(f"  ⚖️ Use PPOCR v4 - Better overall performance")

print(f"\n{'='*60}")
```
